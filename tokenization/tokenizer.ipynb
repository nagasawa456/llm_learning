```python
# 「文章をトークン（単語や記号）に分割し、それぞれをトークンID（整数）に変換する処理（encode）」
# と、その逆の「トークンIDの並びを元の文章に戻す処理（decode）」を行っています。

class SimpleTokenizerV1: # 前後に__があるものはクラスからオブジェクトを作る際に自動で実行される
  def __init__(self, vocab):
    self.str_to_int = vocab
    self.int_to_str = {i:s for s, i in vocab.items()}

  def  encode(self, text):
    preprocessed = re.split(r'([,.?_!"()\']|--|\s)',text)
    preprocessed = [item.strip() for item in preprocessed if item.strip()]
    ids = [self.str_to_int[s] for s in preprocessed] # self.str_to_int[s]は自動で実行されたものを使っていく
    return ids

  def decode(self,ids):
    text = " ".join(([self.int_to_str[i] for i in ids]))
    text = re.sub(r'\s+([,.?!"()\'])', r'\1' , text)
    return text
```

```python
class SimpleTokenizerV2:
  def __init__(self, vocab): # まずは辞書を作る
    self.str_to_int = vocab
    self.int_to_str = { i:s for s, i in vocab.items()}

  def encode(self, text):
      preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
      preprocessed = [
          item.strip() for item in preprocessed if item.strip() # if item.strip()で" "は""となり、item.strip = False となり、リストには追加されない
      ]
      preprocessed = [
          item if item in self.str_to_int  # 辞書にある言葉で～じゃなかったらunkを入れる
          else "<|unk|>" for item in preprocessed
      ]
      ids = [self.str_to_int[s] for s in preprocessed] # idsはトークンID
      return ids
  def decode(self, ids):
    text = " ".join([self.int_to_str[i] for i in ids]) # joinで要素の間に空白を挿入し文章を作っていく
    text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text) # これは正規表現！re.sub(パターン, 置換後, 対象文字列) は、条件（パターン）に合う部分を置き換える関数
                                                      # 置換後の\1は条件に合う部分そのもの（一般化して説明すると、グループ１）のもの
                                                      # 正規表現を書くときは、バックスラッシュの誤作動を防ぐために r'...' と書くのが定石
    return text
```

```python
import torch
from torch.utils.data import Dataset, DataLoader

class GPTDatasetV1(Dataset):
  def __init__(self, txt, tokenizer, max_length, stride):
    self.input_ids = []
    self.target_ids =[]
    token_ids = tokenizer.encode(txt)
    #テキスト全体をトークン化

    #文章をmax_lengthの長さのシーケンスに分割するよ
    for i in range(0, len(token_ids) - max_length, stride):
      input_chunk = token_ids[i:i + max_length]
      target_chunk = token_ids[i + 1: i + max_length + 1]
      #作った入力変数と目的変数をテンソルにするよ
      self.input_ids.append(torch.tensor(input_chunk))
      self.target_ids.append(torch.tensor(target_chunk))

  def __len__(self):
    return len(self.input_ids)
    #データセットに含まれてる行の総数を返す

  def __getitem__(self, idx):
    return self.input_ids[idx], self.target_ids[idx]
    #1行データセットから返す
```

```python
def create_dataloader_v1(txt, batch_size = 4, max_length = 256, stride = 128, shuffle = True, drop_last = True, num_workers = 0):
  tokenizer = tiktoken.get_encoding("gpt2")
  #トークナイザ―を初期化した
  dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)
  #先ほどの処理

  #学習セットは、この「入力を見せて、正解と比べて、パラメータを直す」を何周もするための材料
  #DataLoader はその材料を、モデルが一度に計算できる大きさに切り分けて渡す係です。
  #どれくらいの重さのものを学習材料として与えるかはdataloaderによって決まるよ
  dataloader = DataLoader(
      dataset,
      batch_size = batch_size,
      shuffle = shuffle,
      #drop_last = Trueは最後のバッチが短くなっちゃったらそれを消すための設定
      drop_last = drop_last,
      #num_workersは前処理に使うcpuプロセスの数
      num_workers  = num_workers
  )
  return dataloader
```

```python
with open("the-verdict.txt", "r", encoding = "utf-8") as f:
  raw_text = f.read()

dataloader = create_dataloader_v1(
    raw_text,
    batch_size = 1,
    max_length = 4,
    stride = 1,
    shuffle = False
)
#iterはイテレータというもので、中身を一度に全部抱えず、必要な1個だけ取り出して次の位置を覚えておくための仕組み
data_iter = iter(dataloader)
first_batch = next(data_iter)
print(first_batch)
```

```python
#ここからattentionをやるよ
#まずは訓練可能な重みを一切持たない、単純化されたself-attentionメカニズムを実装する
import torch

inputs = torch.tensor(
    [[0.43, 0.15, 0.89], #Your(x^1)
     [0.55, 0.87, 0.66], #journey(x^2)
     [0.57, 0.85, 0.64], #starts(x^3)
     [0.22, 0.58, 0.33], #with(x`4)
     [0.77, 0.25, 0.10], #one(x^5)
     [0.05, 0.80, 0.55]] #step(x^6)
)
```

```python
query = inputs[1]
#二つ目の入力トークンをクエリとして扱う
#実際には6つのトークンすべてについて計算するよ

attn_scores_2 = torch.empty(inputs.shape[0])
#要素が6つのテンソル(入れ物)を作る。inputs.shape = {6, 3}というイメージ
for i, x_i in enumerate(inputs):
  attn_scores_2[i] = torch.dot(x_i, query)

print(attn_scores_2)
```

```python
#ここではattentinスコアの合計が1になるように合計で割っている
#これが重みの一種であり、練習用の簡易版
#負の数がベクトルに含まれるとうまくいかなくなる
attn_weights_2_tmp = attn_scores_2 / attn_scores_2.sum()
print("Attention weights:", attn_weights_2_tmp)
print("Sum:", attn_weights_2_tmp.sum())
```

```python
def softmax_naive(x):
  return torch.exp(x) / torch.exp(x).sum(dim = 0)

#softmax_naiveはeのスコア乗を計算し、その合計で割るという形
#負のスコアにも対応できる

attn_weights_2_naive = softmax_naive(attn_scores_2)
print("Attention weights:", attn_weights_2_naive)
print("Sum:", attn_weights_2_naive.sum())
```

```python
#性能面で広く最適化されているtorchのソフトマックスを使う
attn_weights_2 = torch.softmax(attn_scores_2, dim = 0)
print("Attention weights:", attn_weights_2)
print("Sum:", attn_weights_2.sum())
```

```python
query = inputs[1]

#求めたattentionスコアと入力トークンを掛け合わせ、それらをすべて足す
#これによってまわりの情報を取り込んで意味を更新することができる
#実際は次元がとてつもなく大きいので、足したとしても区別がちゃんとつく
context_vec_2 = torch.zeros(query.shape)
for i , x_i in enumerate(inputs):
  context_vec_2 += attn_weights_2[i]*x_i

print(context_vec_2)
```

```python
#今までは二つ目の入力に対するattentionの重みとコンテキストベクトルを計算したが、
#これからは、すべての入力に対してattentionの重みとコンテキストベクトルを計算する
#まずはattentionスコアを求める
#dotによる計算は入力トークンとクエリの計算で、3つの要素があるベクトル同士の計算である。
attn_scores = torch.empty(6, 6)
for i, x_i in enumerate(inputs):
  for j , x_j in enumerate(inputs):
    attn_scores[i,j] = torch.dot(x_i, x_j)

print(attn_scores)
```

```python
#forループは遅いので、行列積で求める
attn_scores = inputs @ inputs.T
print(attn_scores)
```

```python
#softmaxで正規化
#dim = -1にすることで、横（行）の合計が1になるようにしてくれる
#最後の次元にどの単語を見るかを設定すれば、dim = -1にすることで見る単語を一つずつ見ていくようになり、
#クエリとすべての単語とのattentionの重みの合計が1になる
attn_weights = torch.softmax(attn_scores, dim = -1)
print(attn_weights)
```

```python
#実際にクエリごとに合計が1になることをかくにんするよ
row_2_sum = sum([0.1385, 0.2379, 0.2333, 0.1240, 0.1082, 0.1581])
print("Row 2 sum:", row_2_sum)
print("All row sums:", attn_weights.sum(dim = -1))
```

```python
#次はクエリごとのattentionの重みを使って、コンテキストベクトルを計算する
#確認:2次元のとき、クエリに対するattentionの重みは見る単語の数だけあって、そのものの次元は1
all_context_vecs = attn_weights @ inputs
print(all_context_vecs)
```

```python
#こっからは訓練可能な重みをもつself-attentionを実装するよ
#今までと同様にまず、一つの入力トークンをクエリとして考えるよ
#重み行列を追加することで学習で調整できる部分を作るようにした(今までは入力ベクトルだけで決まっていた)
#また、関係に対称性があったけれど、今回は単語同士の関係に矢印が追加された

#変数の定義
x_2 = inputs[1]
d_in = inputs.shape[1]
d_out = 2
#普通は入力と出力の次元は一緒だけど計算を追いやすくするために変えてるよ

#3つの重み行列の初期化
torch.manual_seed(123)
W_query = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad = False)
W_key = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad = False)
W_value = torch.nn.Parameter(torch.rand(d_in, d_out), requires_grad = False)
#普通はrequires_gradはtrue

#ベクトルの計算
query_2 = x_2 @ W_query
key_2 = x_2 @ W_key
value_2 = x_2 @ W_value
print(query_2) # 重み行列の列数を2にしたため、出力は2次元ベクトル
```

```python
#まとめてキーベクトル、値ベクトルを求めている
keys = inputs @ W_key
values = inputs @ W_value
print("keys.shape", keys.shape)
print("values.shape", values.shape)
# 行列積を用いて求める
```

```python
#次はattentionスコアを計算していくよ
keys_2 = keys[1]
#キーベクトルとクエリベクトルをドット積で計算する
attn_score_22 = query_2.dot(keys_2)
print(attn_score_22)
```

```python
#あるクエリに対するすべてのattentionスコアを計算する
attn_scores_2 = query_2 @ keys.T
print(attn_scores_2)
```

```python
#次はattentionスコアからattentionの重みを計算する
d_k = keys.shape[-1]
attn_weights_2 = torch.softmax(attn_scores_2 / d_k**0.5, dim = -1)
print(attn_weights_2)
```

```python
# GPTモデル全体のデータの流れ（トークン入力から確率計算まで）を確認するために、
# 未実装の複雑なパーツを中身が空の仮部品（ダミー）で置き換えて作った
# GPTモデルの基本骨格（パイプライン）
import torch
import torch.nn as nn

class DummyGPTModel(nn.Module):

  def __init__(self, cfg):
    super().__init__()
    self.tok_emb = nn.Embedding(cfg["vocab_size"], cfg["emb_dim"])# 単語IDをemb_dimの意味ベクトルに変換する
    self.pos_emb = nn.Embedding(cfg["context_length"], cfg["emb_dim"])# 文章の何番目にある単語かを示す位置ベクトルを生成する
    self.drop_emb = nn.Dropout(cfg["drop_rate"])#

    self.trf_blocks = DummyLayerNorm(cfg["emb_dim"]) # 本来はSelf-Attentionなどが動くメインの処理層で、今はDummyLayerNormで素通りさせている
    self.out_head = nn.Linear(
        cfg["emb_dim"], cfg["vocab_size"], bias = False
    )

  def forward(self, in_idx):
    batch_size, seq_len = in_idx.shape
    tok_embeds = self.tok_emb(in_idx)
    pos_embeds = self.pos_emb(
        torch.arange(seq_len, device = in_idx.device)
    )
    x = tok_embeds + pos_embeds # 単語の意味と位置の情報を足し合わせる
    x = self.drop_emb(x)
    x = self.trf_blocks(x)
    logits = self.out_head(x)
    return logits

class DummyTransformerBlock(nn.Module):
  def __init__(self, cfg):
    super().__init__()

  def forward(self, x):
    return x

class DummyLayerNorm(nn.Module):

  def __init__(self, normalized_shape, eps = 1e-5):
    super().__init__()

  def forward(self, x):
    return x
```
